```python
clean_df = df.copy()

print("Original shape:", df.shape)
print("Working dataset shape:", clean_df.shape)
```

```python
clean_df = clean_df.drop_duplicates()

print("Shape after removing duplicates:", clean_df.shape)
```

```python
numeric_columns = [
    "Age",
    "Booking_Count",
    "Booking_Amount",
    "Trip_Duration",
    "Customer_Rating",
    "Customer_Likes",
    "Reviews",
    "Hotel_Rating",
    "Cancellation_Count"
]

for column in numeric_columns:
    clean_df[column] = pd.to_numeric(
        clean_df[column],
        errors="coerce"
    )
```

```python
clean_df.loc[
    (clean_df["Age"] < 18) | (clean_df["Age"] > 100),
    "Age"
] = np.nan
```

```python
clean_df.loc[
    (clean_df["Customer_Rating"] < 0) |
    (clean_df["Customer_Rating"] > 5),
    "Customer_Rating"
] = np.nan
```

```python
clean_df.loc[
    (clean_df["Hotel_Rating"] < 0) |
    (clean_df["Hotel_Rating"] > 5),
    "Hotel_Rating"
] = np.nan
```

```python
clean_df.loc[
    clean_df["Booking_Amount"] < 0,
    "Booking_Amount"
] = np.nan
```

```python
clean_df.loc[
    clean_df["Trip_Duration"] <= 0,
    "Trip_Duration"
] = np.nan
```

```python
for column in numeric_columns:
    clean_df[column] = clean_df[column].fillna(
        clean_df[column].median()
    )
```

```python
clean_df.to_csv(
    "cleaned_travel_dataset.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
